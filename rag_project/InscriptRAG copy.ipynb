```python
from langchain_community.document_loaders import DirectoryLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_huggingface import HuggingFaceEmbeddings

class RagRetriever:
    """
    Builds vector store ONCE and reuses it
    """

    def __init__(self, top_k: int):
        self.top_k = top_k
        self.vectorstore = None

    def load_dataset(self):
        loader = DirectoryLoader("Dataset")
        return loader.load()

    def chunk_documents(self, documents):
        splitter = RecursiveCharacterTextSplitter(
            chunk_size=500,
            chunk_overlap=100
        )
        return splitter.split_documents(documents)

    def build_vectorstore(self):
        documents = self.load_dataset()
        chunks = self.chunk_documents(documents)

        self.vectorstore = Chroma.from_documents(
            documents=chunks,
            embeddings = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2"
)
,
            persist_directory="./chroma_db"
        )

        self.vectorstore.persist()
        print("✅ Vectorstore built and persisted")

    def get_retriever(self):
        if self.vectorstore is None:
            raise ValueError("Vectorstore not built. Call build_vectorstore() first.")

        return self.vectorstore.as_retriever(
            search_kwargs={"k": self.top_k}
        )
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI


class LLM:
    """
    Handles all LLM interactions:
    - Entity extraction (NER)pip install sentence-transformers
    - Answer generation
    """

    def __init__(self):
        self.model = ChatGoogleGenerativeAI(
            model="gemini-2.5-flash",
            temperature=0
        )

        self.ner_template = """
        Given the following text.
        Identify a maximum of 3 most important entities (proper nouns such as names, locations, objects).
        You may identify fewer than three if not applicable.
        Do not include question words, verbs, or prepositions as entities.
        Return the output strictly in this format:
        entity1;entity2;entity3

        Text: {text}
        """

        self.gen_ans_template = """
        You are an epigraphist answering questions on Indian inscriptions.
        Use only the provided inscription extract to answer the question.
        If the answer is not present, say that you do not know.

        Question: {question}
        Context: {context}
        Answer:
        """

    def invoke_llm_ner(self, query: str):
        """
        Extract key entities from the query using Gemini
        """
        prompt = ChatPromptTemplate.from_template(self.ner_template)
        parser = StrOutputParser()

        chain = prompt | self.model | parser

        response = chain.invoke({"text": query})
        print("--> NER successful")

        entities = [e.strip() for e in response.split(";") if e.strip()]

        print("NER", len(entities), entities)
        return entities

    def extract_entities(self, query: str):
        """
        Return query + extracted entities for retrieval
        """
        entities = self.invoke_llm_ner(query)
        return [query] + entities

    def get_answer(self, query: str, context: str):
        """
        Generate final answer using retrieved context
        """
        prompt = ChatPromptTemplate.from_template(self.gen_ans_template)
        parser = StrOutputParser()

        chain = prompt | self.model | parser

        answer = chain.invoke({
            "question": query,
            "context": context
        })

        print("--> Generating answer successful")
        return answer
```

```python
class GetContexts:

  def __init__(self, query_entities, documents):
    self.query_entities = query_entities
    self.documents = documents

  def retrive_contexts(self, retriever, entity):
    return [docs.metadata for docs in retriever.get_relevant_documents(entity)]

  def collate_contexts(self, retriever):
    contexts_set = set()
    for entity in self.query_entities:
        contexts = self.retrive_contexts(retriever, entity)
        for c in contexts:
            contexts_set.add(c['source'])
    return contexts_set

  def get_contexts_documents(self, retriever):
    context_documents = ""
    contexts_list = []
    contexts_set = self.collate_contexts(retriever)
    print(contexts_set)
    for doc in self.documents:
      if doc.metadata['source'] in contexts_set:
        #print(doc.page_content)
        context_documents = context_documents + doc.page_content
        contexts_list.append(doc.page_content)
    return context_documents, contexts_list
```

```python
import json

class TheClass:

  def __init__(self):
    self.queries_path = 'questions.txt'
    self.contexts_path = 'contexts.json'
    self.answers_path = 'answers.json'

  def load_queries(self):
    with open(self.queries_path) as f:
      queries = f.readlines()

      return queries

  def write_to_json(self, path, list_file):
    with open(path, "w", encoding="utf-8") as f:
      f.write(json.dumps(list_file, indent=2))

  def run_one_query(self, query):

    llm = LLM()

    rag = RagRetriever(top_k=1)
    rag.build_vectorstore()
    retriever = rag.get_retriever()

    documents = rag.load_dataset()

    query_entities = llm.extract_entities(query)

    get_contexts = GetContexts(query_entities, documents)
    context_documents, _ = get_contexts.get_contexts_documents(retriever)

    answer = llm.get_answer(query, context_documents)

    return answer


  def run_the_entire_thing(self):
    """
    """
    queries = self.load_queries()
    contexts_total = []
    answers_total = []
    for query in queries:
      print("\n =============== \nQuery:", query)
      llm = LLM()
      query_entities = llm.extract_entities(query)
      rag_retriever = RagRetriever(1)
      documents = rag_retriever.load_dataset()
      get_contexts = GetContexts(query_entities, documents)
      context_documents, contexts_list = get_contexts.get_contexts_documents(rag_retriever.retrieve())
      contexts_total.append(contexts_list)
      #print("Contexts: ", contexts_list)
      answer = llm.get_answer(query, context_documents)
      #print("Answer: ", answer)
      answers_total.append(answer)
    self.write_to_json(self.contexts_path, contexts_total)
    self.write_to_json(self.answers_path, answers_total)

    return contexts_total , answers_total

```

```python
c = TheClass()
print(c.run_one_query("Who was Ashoka?"))
```

```python
import json

def load_json(file_name):
  with open(file_name, encoding="utf-8") as f:
      response = json.load(f)
      return response

def load_text(file_name):
    with open(file_name, encoding="utf-8") as f:
      response = f.readlines()
      return response


contexts = load_json('contexts.json')
answers = load_json('answers.json')
ground_truths = load_text('ground_truth.txt')
questions = load_text('questions.txt')
```

```python
import pandas as pd

df = pd.DataFrame()

df['question'] = questions
df['answer'] = answers
df['contexts'] = contexts
df['ground_truth'] = ground_truths
```

```python
from ragas import evaluate
from ragas.metrics import answer_correctness
from datasets import Dataset

dataset = Dataset.from_pandas(df)
answer_correctness = evaluate(dataset,metrics=[answer_correctness])
print(answer_correctness.to_pandas())
```

```python
from deepeval.metrics.ragas import (
    RAGASContextualPrecisionMetric,
    RAGASFaithfulnessMetric,
    RAGASContextualRecallMetric,
    RAGASAnswerRelevancyMetric,
)

contextual_precision = RAGASContextualPrecisionMetric()
contextual_recall = RAGASContextualRecallMetric()
answer_relevancy = RAGASAnswerRelevancyMetric()
faithfulness = RAGASFaithfulnessMetric()
```

```python
from deepeval import evaluate
from deepeval.test_case import LLMTestCase

test_case = LLMTestCase(
  input=questions[0],
  actual_output=answers[0],
  expected_output=ground_truths[0],
  retrieval_context=contexts[0]
)


evaluate(
  test_cases=[test_case],
  metrics=[
    contextual_precision,
    contextual_recall,
    answer_relevancy,
    faithfulness,
  ]
)
```
